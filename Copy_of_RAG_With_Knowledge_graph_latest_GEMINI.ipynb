````python
def generate_full_text_query(input: str) -> str:
    full_text_query = ""
    words = [el for el in remove_lucene_chars(input).split() if el]
    for word in words[:-1]:
        full_text_query += f" {word}~2 AND"
    full_text_query += f" {words[-1]}~2"
    return full_text_query.strip()
````

````python
# Fulltext index query
def structured_retriever(question: str) -> str:
    result = ""
    entities = entity_chain.invoke({"question": question})
    for entity in entities.names:
        response = graph.query(
            """CALL db.index.fulltext.queryNodes('entity', $query, {limit:2})
            YIELD node,score
            CALL {
              WITH node
              MATCH (node)-[r:!MENTIONS]->(neighbor)
              RETURN node.id + ' - ' + type(r) + ' -> ' + neighbor.id AS output
              UNION ALL
              WITH node
              MATCH (node)<-[r:!MENTIONS]-(neighbor)
              RETURN neighbor.id + ' - ' + type(r) + ' -> ' +  node.id AS output
            }
            RETURN output LIMIT 50
            """,
            {"query": generate_full_text_query(entity)},
        )
        result += "\n".join([el['output'] for el in response])
    return result
````

````python
def retriever(question: str):
    print(f"Search query: {question}")
    structured_data = structured_retriever(question)
    unstructured_data = [el.page_content for el in vector_index.similarity_search(question)]
    final_data = f"""Structured data:
{structured_data}
Unstructured data:
{"#Document ". join(unstructured_data)}
    """
    return final_data
````

````python
def _format_chat_history(chat_history: List[Tuple[str, str]]) -> List:
    buffer = []
    for human, ai in chat_history:
        buffer.append(HumanMessage(content=human))
        buffer.append(AIMessage(content=ai))
    return buffer
````

````python
_search_query = RunnableBranch(
    # If input includes chat_history, we condense it with the follow-up question
    (
        RunnableLambda(lambda x: bool(x.get("chat_history"))).with_config(
            run_name="HasChatHistoryCheck"
        ),  # Condense follow-up question and chat into a standalone_question
        RunnablePassthrough.assign(
            chat_history=lambda x: _format_chat_history(x["chat_history"])
        )
        | CONDENSE_QUESTION_PROMPT
        | llm
        | StrOutputParser(),
    ),
    # Else, we have no chat history, so just pass through the question
    RunnableLambda(lambda x : x["question"]),
)
````

## Evaluation Metrics for RAG Chain

To effectively evaluate a Retrieval-Augmented Generation (RAG) pipeline, you typically need a comprehensive dataset that includes questions and their corresponding ground-truth answers. Optionally, you might also include ground-truth contexts if you want to evaluate the retrieval component separately.

The process generally involves:
1.  **Defining an Evaluation Dataset**: A collection of `(question, ground_truth_answer)` pairs.
2.  **Running the RAG Chain**: For each question in the dataset, invoke your `chain` to get the generated answer.
3.  **Calculating Metrics**: Compare the generated answers against the ground-truth answers using various metrics.

### Important Note on API Quota

````python
# Install the ROUGE scoring library if you plan to use it
!pip install rouge-score

from typing import List, Dict
from rouge_score import rouge_scorer # Uncomment if using rouge-score

# --- Step 1: Define an Evaluation Dataset (Example) ---
# In a real scenario, this would be loaded from a file (CSV, JSON) or a database.
eval_data: List[Dict[str, str]] = [
    {
        "question": "When was Elizabeth I born?",
        "ground_truth_answer": "Elizabeth I was born on 7 September 1533."
    },
    {
        "question": "Who was Elizabeth I's mother?",
        "ground_truth_answer": "Elizabeth I's mother was Anne Boleyn."
    },
    {
        "question": "What was one of Elizabeth I's mottoes?",
        "ground_truth_answer": "One of her mottoes was 'video et taceo' (I see and keep silent)."
    }
]

# --- Step 2: Run the RAG Chain for each question ---
predictions: List[Dict[str, str]] = []

print("Starting RAG chain invocation for evaluation data...")
for i, entry in enumerate(eval_data):
    question = entry["question"]
    print(f"Processing question {i+1}/{len(eval_data)}: {question}")
    try:
        # Invoke the RAG chain. Ensure your LLM quota is sufficient.
        generated_answer = chain.invoke({"question": question, "chat_history": []})
        predictions.append({
            "question": question,
            "generated_answer": generated_answer,
            "ground_truth_answer": entry["ground_truth_answer"]
        })
        print(f"  Generated: {generated_answer}\n")
    except Exception as e:
        print(f"  Error generating answer for '{question}': {e}\n")
        predictions.append({
            "question": question,
            "generated_answer": f"Error: {e}", # Record the error instead of an answer
            "ground_truth_answer": entry["ground_truth_answer"]
        })

print("RAG chain invocation complete.")

# --- Step 3: Placeholder for Calculating Metrics ---
print("\n--- Calculating Evaluation Metrics ---")

# Example Metric 1: Exact Match Score
def calculate_exact_match(predictions_list: List[Dict[str, str]]) -> float:
    correct_matches = 0
    for p in predictions_list:
        if "Error" not in p["generated_answer"] and \
           p["generated_answer"].strip().lower() == p["ground_truth_answer"].strip().lower():
            correct_matches += 1
    return correct_matches / len(predictions_list) if predictions_list else 0.0

exact_match_score = calculate_exact_match(predictions)
print(f"Exact Match Score: {exact_match_score:.2f}")

# Example Metric 2: ROUGE Score (requires 'rouge-score' library)
# ROUGE (Recall-Oriented Understudy for Gisting Evaluation) is a set of metrics
# for evaluating automatic summarization and machine translation.
# It compares an automatically produced summary or translation against a reference or set of references.
# This part is commented out as it requires an external library installation.
if 'rouge_scorer' in locals(): # Check if rouge_scorer was imported
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rougeL'], use_stemmer=True)
    rouge_l_f1_scores = []
    for p in predictions:
        if "Error" not in p["generated_answer"]:
            scores = scorer.score(p['ground_truth_answer'], p['generated_answer'])
            rouge_l_f1_scores.append(scores['rougeL'].fmeasure)
    if rouge_l_f1_scores:
        average_rouge_l = sum(rouge_l_f1_scores) / len(rouge_l_f1_scores)
        print(f"Average ROUGE-L F1-Score: {average_rouge_l:.2f}")
    else:
        print("Could not calculate ROUGE-L F1-Score (no valid generated answers).")
else:
    print("ROUGE-L F1-Score calculation skipped (rouge-score library not installed or imported).")

# Other potential metrics to consider:
# - Semantic Similarity (e.g., using embeddings to compare generated and ground-truth answers)
# - Faithfulness/Factuality (does the generated answer contain information not present in the context or is it hallucinated?)
# - Answer Relevance (is the generated answer relevant to the question?)
# - Context Relevance (is the retrieved context relevant to the question?)
# - Context Recall (does the retrieved context contain all necessary information to answer the question?)

print("\n--- Raw Predictions for detailed review ---")
for pred in predictions:
    print(f"Question: {pred['question']}")
    print(f"  Generated Answer: {pred['generated_answer']}")
    print(f"  Ground Truth: {pred['ground_truth_answer']}")
    print("------------------------------------------")

print("\nEvaluation framework setup complete. Implement more advanced metrics as needed.")
````
